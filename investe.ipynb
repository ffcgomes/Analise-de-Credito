```python
def atualizar_saldos(investimentos):
    for investimento in investimentos:
        print(f"Investimento: {investimento.nome}")
        saldo_atual = float(input("Digite o saldo atual: "))
        investimento.saldo_atual = saldo_atual
```

```python
def exibir_investimentos(investimentos):
    data_atual = datetime.date.today()
    for investimento in investimentos:
        print(f"Investimento: {investimento.nome}")
        print(f"Data do investimento: {investimento.data_investimento.strftime('%d/%m/%Y')}")
        print(f"Valor: {investimento.valor}")
        print(f"Taxa contratada: {investimento.taxa_contratada}")
        print(f"Data de vencimento: {investimento.data_vencimento.strftime('%d/%m/%Y')}")
        print(f"Saldo atual: {investimento.saldo_atual}")
        print(f"Lucro real: {investimento.calcular_lucro_real()}")
        print(f"Taxa de juros real: {investimento.calcular_taxa_juros_real()}")
        print("-" * 20)

    # Criação do gráfico e da tabela

    # Cria um DataFrame com os dados dos investimentos
    dados_grafico = []
    for investimento in investimentos:
        dados_grafico.append([investimento.nome, investimento.valor, investimento.saldo_atual])
    df_grafico = pd.DataFrame(dados_grafico, columns=['Investimento', 'Valor Inicial', 'Saldo Atual'])

    # Cria o gráfico de barras
    df_grafico.plot(x='Investimento', y=['Valor Inicial', 'Saldo Atual'], kind='bar')
    plt.title('Evolução dos Investimentos')
    plt.ylabel('Valor (R$)')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

    # Cria a tabela com os dados dos investimentos
    display(df_grafico)
```

```python
atualizar_saldos(investimentos)
```
